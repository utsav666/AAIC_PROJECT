````python
# Cell 3: Prompt Engineering approach (baseline)

from openai import OpenAI
import json

client = OpenAI(api_key=os.environ["OPENAI_API_KEY"])

# Long prompt with examples (few-shot)
FEW_SHOT_PROMPT = """You are a Merck Service Agreement data extractor. 
Extract structured data from service agreements and return ONLY valid JSON.

The output must follow this exact schema:
{
    "staffing_table": [
        {"level": "<role/title>", "site": "<location>", "base_charges": "<rate>"}
    ],
    "contact_information": {
        "merck_office_phone": "<phone>",
        "merck_email": "<email>",
        "supplier_office_phone": "<phone>",
        "supplier_email": "<email>"
    }
}

IMPORTANT RULES:
1. Map any role/title/position/grade/designation field to "level"
2. Map any location/site/base/office field to "site"
3. Map any rate/fee/charge/cost/billing/price field to "base_charges"
4. Map Merck phone variations (Tel, Phone, Landline, Direct Line, #) to "merck_office_phone"
5. Map Merck email variations (Email, Mail, Electronic Address, @) to "merck_email"
6. Map Supplier/Vendor/Partner phone to "supplier_office_phone"
7. Map Supplier/Vendor/Partner email to "supplier_email"
8. Return ONLY JSON, no explanations

EXAMPLE 1:
Input: "STAFFING TABLE\\nLevel: Senior Developer\\nSite: Rahway, NJ\\nBase Charges: $150/hr\\n\\nCONTACT INFORMATION\\nMerck Office Phone: 908-555-1234\\nMerck Email: contracts@merck.com\\nSupplier Office Phone: 212-555-5678\\nSupplier Email: support@vendor.com"

Output: {"staffing_table":[{"level":"Senior Developer","site":"Rahway, NJ","base_charges":"$150/hr"}],"contact_information":{"merck_office_phone":"908-555-1234","merck_email":"contracts@merck.com","supplier_office_phone":"212-555-5678","supplier_email":"support@vendor.com"}}

EXAMPLE 2:
Input: "PERSONNEL CHARGES\\nGrade: Principal Consultant\\nBase: Zurich, Switzerland\\nFee: CHF 250/hr\\n\\nCOMMUNICATION\\nMerck Telephone: +41-44-555-1111\\nMerck E-mail: emea-sourcing@merck.com\\nVendor Telephone: +49-89-555-2222\\nVendor E-mail: consulting@eupartner.de"

Output: {"staffing_table":[{"level":"Principal Consultant","site":"Zurich, Switzerland","base_charges":"CHF 250/hr"}],"contact_information":{"merck_office_phone":"+41-44-555-1111","merck_email":"emea-sourcing@merck.com","supplier_office_phone":"+49-89-555-2222","supplier_email":"consulting@eupartner.de"}}

Now extract from the following:
"""

def extract_with_prompting(text: str) -> dict:
    """Extract using few-shot prompting (no fine-tuning)"""
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",  # or gpt-3.5-turbo
        messages=[
            {"role": "user", "content": FEW_SHOT_PROMPT + text}
        ],
        temperature=0,
    )
    raw = response.choices[0].message.content.strip()
    # Clean markdown if present
    if raw.startswith("```"):
        raw = raw.split("```")[1]
        if raw.startswith("json"):
            raw = raw[4:]
    try:
        return json.loads(raw)
    except json.JSONDecodeError:
        return {"error": "Failed to parse JSON", "raw": raw}

# Test prompt engineering approach
print("=" * 60)
print("PROMPT ENGINEERING RESULTS")
print("=" * 60)

prompt_results = []
for tc in TEST_CASES:
    result = extract_with_prompting(tc["input"])
    prompt_results.append({
        "id": tc["id"],
        "result": result,
        "expected": tc["expected"],
        "match": result == tc["expected"]
    })
    print(f"\n{tc['id']}:")
    print(f"Match: {result == tc['expected']}")
    print(f"Result: {json.dumps(result, indent=2)}")
````

````python
# Cell 4: Upload file and create fine-tuning job

from openai import OpenAI

client = OpenAI(api_key=os.environ["OPENAI_API_KEY"])

# Upload training file
training_file = client.files.create(
    file=open("merck_service_agreement_training.jsonl", "rb"),
    purpose="fine-tune",
)
print("training_file_id:", training_file.id)

# Create fine-tuning job
# Replace BASE_MODEL with a fine-tune-capable model from your org
BASE_MODEL = "gpt-3.5-turbo"  # e.g., "gpt-4o-mini", check your dashboard
SUFFIX = "merck-extractor"

job = client.fine_tuning.jobs.create(
    model=BASE_MODEL,
    training_file=training_file.id,
    suffix=SUFFIX,
)

print("job_id:", job.id)
print("status:", job.status)
````

````python
# Cell 5: Wait for fine-tuning to complete

import time

terminal = {"succeeded", "failed", "cancelled"}

while True:
    job = client.fine_tuning.jobs.retrieve(job.id)
    print("status:", job.status)

    if job.status in terminal:
        break

    time.sleep(30)

print("=" * 40)
print("final_status:", job.status)

if job.status == "succeeded":
    fine_tuned_model = job.fine_tuned_model
    print("fine_tuned_model:", fine_tuned_model)
else:
    print("error:", getattr(job, "error", None))
    # List events for debugging
    events = client.fine_tuning.jobs.list_events(job.id, limit=20)
    for e in reversed(events.data):
        print(f"{e.created_at}: {e.message}")
````

````python
# Cell 6: Extract using fine-tuned model

# Set this after job succeeds
FINE_TUNED_MODEL = job.fine_tuned_model  # or paste the model ID manually

def extract_with_finetuned(text: str) -> dict:
    """Extract using fine-tuned model (minimal prompt)"""
    response = client.chat.completions.create(
        model=FINE_TUNED_MODEL,
        messages=[
            {"role": "system", "content": "You are a Merck Service Agreement data extractor. Extract structured data as JSON only."},
            {"role": "user", "content": text}
        ],
        temperature=0,
    )
    raw = response.choices[0].message.content.strip()
    try:
        return json.loads(raw)
    except json.JSONDecodeError:
        return {"error": "Failed to parse JSON", "raw": raw}

# Test fine-tuned approach
print("=" * 60)
print("FINE-TUNED MODEL RESULTS")
print("=" * 60)

finetuned_results = []
for tc in TEST_CASES:
    result = extract_with_finetuned(tc["input"])
    finetuned_results.append({
        "id": tc["id"],
        "result": result,
        "expected": tc["expected"],
        "match": result == tc["expected"]
    })
    print(f"\n{tc['id']}:")
    print(f"Match: {result == tc['expected']}")
    print(f"Result: {json.dumps(result, indent=2)}")
````

````python
# Cell 7: Comparison summary

import tiktoken

enc = tiktoken.encoding_for_model("gpt-4o-mini")

# Token counts
prompt_tokens = len(enc.encode(FEW_SHOT_PROMPT))
finetuned_system_prompt = "You are a Merck Service Agreement data extractor. Extract structured data as JSON only."
finetuned_tokens = len(enc.encode(finetuned_system_prompt))

print("=" * 70)
print("COMPARISON: PROMPT ENGINEERING vs FINE-TUNING")
print("=" * 70)

print("\n1. PROMPT SIZE")
print(f"   Few-shot prompt tokens:    {prompt_tokens}")
print(f"   Fine-tuned system tokens:  {finetuned_tokens}")
print(f"   Token savings per call:    {prompt_tokens - finetuned_tokens} ({100*(prompt_tokens - finetuned_tokens)/prompt_tokens:.1f}%)")

print("\n2. ACCURACY")
prompt_correct = sum(1 for r in prompt_results if r["match"])
finetuned_correct = sum(1 for r in finetuned_results if r["match"])
print(f"   Prompt engineering:        {prompt_correct}/{len(TEST_CASES)} correct")
print(f"   Fine-tuned model:          {finetuned_correct}/{len(TEST_CASES)} correct")

print("\n3. CONSISTENCY")
print("   Prompt engineering:        Variable (depends on prompt quality)")
print("   Fine-tuned model:          High (learned patterns from examples)")

print("\n4. MAINTENANCE")
print("   Prompt engineering:        Update prompt text, redeploy")
print("   Fine-tuned model:          Retrain with new examples")

print("\n5. COST PER CALL")
print(f"   Prompt engineering:        ~{prompt_tokens} input tokens/call")
print(f"   Fine-tuned model:          ~{finetuned_tokens} input tokens/call + training cost (one-time)")

print("\n" + "=" * 70)
print("RECOMMENDATION")
print("=" * 70)
print("""
Use FINE-TUNING when:
  - You have many similar extraction tasks (high volume)
  - Output format must be strictly consistent
  - Domain terminology varies (Level vs Grade vs Role vs Position)
  - You want to reduce per-call token costs

Use PROMPT ENGINEERING when:
  - Low volume / occasional use
  - Task requirements change frequently
  - You can't create 10+ training examples
  - Quick iteration is more important than consistency
""")
````
